## Classes

A Python *class* is a code block that defines a custom object and determines its behavior; the `class` keyword defines and names it. Every class needs a *constructor*, always named `__init__()`, which is called whenever the class instantiates a new object and sets its initial state.

An *attribute* is a variable stored within an object, assigned and accessed in the class body via the name `self`, which refers to the object internally. A function attached to a class is a *method*. The first argument of each method must be `self`; it is included in the declaration but not when calling the method on an instance (`my_backpack.put("notebook")` is equivalent to `Backpack.put(my_backpack, "notebook")`).

### Instantiation

The class block is only a blueprint. "Calling" the class name like a function triggers the constructor and returns a new *instance*, an object whose type is the class. The spec's starting `Backpack` (just a name and a list of contents), under a separate name here so it doesn't clash with the full solution below:

````python
class SimpleBackpack:
    """A Backpack object class. Has a name and a list of contents.

    Attributes:
        name (str): the name of the backpack's owner.
        contents (list): the contents of the backpack.
    """
    def __init__(self, name):           # This function is the constructor.
        """Set the name and initialize an empty list of contents.

        Parameters:
            name (str): the name of the backpack's owner.
        """
        self.name = name                # Initialize some attributes.
        self.contents = []

    def put(self, item):
        """Add 'item' to the backpack's list of contents."""
        self.contents.append(item)      # Use 'self.contents', not just 'contents'.

    def take(self, item):
        """Remove 'item' from the backpack's list of contents."""
        self.contents.remove(item)
````

````python
my_backpack = SimpleBackpack("Fred")
print(type(my_backpack))
print(my_backpack.name, my_backpack.contents)
my_backpack.name = "George"             # Attributes can be modified dynamically.
my_backpack.put("notebook")
my_backpack.put("pencils")
my_backpack.take("pencils")
print(my_backpack.name, my_backpack.contents)
````

If the module is being run directly, not imported, `__name__` is set to `"__main__"`, so commands under an `if __name__ == "__main__":` clause are ignored when the module is imported.

The full class is below, including the `__eq__()` and `__str__()` magic methods from Problem 3 (the magic methods are explained further down).

````python
class Backpack:
    """A Backpack object class. Has a name and a list of contents.

    Attributes:
        name (str): the name of the backpack's owner.
        contents (list): the contents of the backpack.
        color (str): the color of the backpack.
        max_size (int): the capacity of the contents.
    """

    # Problem 1: Modify __init__() and put(), and write dump().
    def __init__(self, name, color, max_size=5):
        """Set the name and initialize an empty list of contents.

        Parameters:
            name (str): the name of the backpack's owner.
            color (str): the color of the backpack.
            max_size (int): the capacity of the contents. Default ot 5
        """
        self.name = name
        self.contents = []
        self.color = color
        self.max_size = max_size

    def put(self, item):
        """Add 'item' to the backpack's list of contents.
           If the number of contents exceeds the size capacity,
           print "No Room!"
        """
        if len(self.contents) < self.max_size:
            self.contents.append(item)
        else:
            print("No Room!")

    def take(self, item):
        """Remove 'item' from the backpack's list of contents."""
        self.contents.remove(item)

    def dump(self):
        """Resets the contents of the backpack"""
        self.contents = []

    # Magic Methods -----------------------------------------------------------

    # Problem 3: Write __eq__() and __str__().
    def __add__(self, other):
        """Add the number of contents of each Backpack."""
        return len(self.contents) + len(other.contents)

    def __lt__(self, other):
        """Compare two backpacks. If 'self' has fewer contents
        than 'other', return True. Otherwise, return False.
        """
        return len(self.contents) < len(other.contents)
    def __eq__(self, other):
        return self.name == other.name and self.color == other.color and len(self.contents) == len(other.contents)

    def __str__(self):
        s = "Owner:\t\t" + self.name + "\n"
        s += "Color:\t\t" + self.color + "\n"
        s += "Size:\t\t" + str(len(self.contents)) + "\n"
        s += "Max Size:\t" + str(self.max_size) + "\n"
        s += "Contents:\t" + str(self.contents)
        return s
````

````python
def test_prob1():
    testpack = Backpack("Barry", "black") # Instantiate the object.
    if testpack.name != "Barry": # Test an attribute.
        print("Backpack.name assigned incorrectly")
    for item in ["pencil", "pen", "paper", "computer", "wallet"]:
        testpack.put(item) # Test a method.
    print("Name:", testpack.name)
    print("Color:", testpack.color)
    print("Contents:", testpack.contents)
    testpack.put("calculator")
    print("Contents:", testpack.contents)
    testpack.dump()
    print("Contents:", testpack.contents)

test_prob1()
````

## Inheritance

To create a class similar to an existing one, it is often better to *inherit* from it than to start from scratch. A class that inherits from another is a *subclass*; the class it inherits from is its *superclass*. To define a subclass, put the superclass name in parentheses after the class name.

A subclass may add new attributes and methods, and can *override* superclass methods by defining them again. The provided `Knapsack` class (which the lab does not require modifying) is a smaller backpack that can be tied closed:

````python
# An example of inheritance. You are not required to modify this class.
class Knapsack(Backpack):
    """A Knapsack object class. Inherits from the Backpack class.
    A knapsack is smaller than a backpack and can be tied closed.

    Attributes:
        name (str): the name of the knapsack's owner.
        color (str): the color of the knapsack.
        max_size (int): the maximum number of items that can fit inside.
        contents (list): the contents of the backpack.
        closed (bool): whether or not the knapsack is tied shut.
    """
    def __init__(self, name, color):
        """Use the Backpack constructor to initialize the name, color,
        and max_size attributes. A knapsack only holds 3 item by default.

        Parameters:
            name (str): the name of the knapsack's owner.
            color (str): the color of the knapsack.
            max_size (int): the maximum number of items that can fit inside.
        """
        Backpack.__init__(self, name, color, max_size=3)
        self.closed = True

    def put(self, item):
        """If the knapsack is untied, use the Backpack.put() method."""
        if self.closed:
            print("I'm closed!")
        else:
            Backpack.put(self, item)

    def take(self, item):
        """If the knapsack is untied, use the Backpack.take() method."""
        if self.closed:
            print("I'm closed!")
        else:
            Backpack.take(self, item)

    def weight(self):
        """Calculate the weight of the knapsack by counting the length of the
        string representations of each item in the contents list.
        """
        return sum([len(str(item)) for item in self.contents])
````

Since `Knapsack` inherits from `Backpack`, a knapsack object *is* a backpack object, and every `Backpack` method (like `dump()`) is available to it. `issubclass()` shows whether one class is derived from another, `isinstance()` whether an object belongs to a class hierarchy, and `hasattr()` whether a class or object has a given attribute or method.

````python
my_knapsack = Knapsack("Brady", "brown")
my_backpack = Backpack("Fred", "red")
print(issubclass(Knapsack, Backpack), issubclass(Backpack, Knapsack))
print(isinstance(my_knapsack, Knapsack) and isinstance(my_knapsack, Backpack))

my_knapsack.put("compass")              # put() and take() need it open.
my_knapsack.closed = False
my_knapsack.put("compass")
my_knapsack.put("pocket knife")
print(my_knapsack.contents, my_knapsack.weight())
print(hasattr(my_knapsack, 'weight'), hasattr(my_backpack, 'weight'))
my_knapsack.dump()                      # Inherited from Backpack.
print(my_knapsack.contents)
````

## Problem 2: `Jetpack`

> **Problem 2.** Write a `Jetpack` class that inherits from the `Backpack` class.
> 1. Override the constructor so that in addition to a name, color, and maximum size, it also accepts an amount of fuel. Change the default value of `max_size` to 2, and set the default value of `fuel` to 10. Store the fuel as an attribute.
> 2. Add a `fly()` method that accepts an amount of fuel to be burned and decrements the `fuel` attribute by that amount. If the user tries to burn more fuel than remains, print "Not enough fuel!" and do not decrement the fuel.
> 3. Override the `dump()` method so that both the contents and the fuel tank are emptied.
> 4. Write clear, detailed docstrings for the class and each of its methods.

````python
class Jetpack(Backpack):
    """A Jetpack object class. Inherits from the Backpack class.
    A jetpack is smaller than a backpack, and can fly using fuel.

    Attributes:
        name (str): the name of the jetpack's owner.
        color(str): the color of the jetpack.
        max_size (int): the maximum number of items that can fit inside.
        contents (list): the contents of the jetpack.
        fuel (int): the amount of fuel left in the jetpack.
    """
    def __init__(self, name, color, max_size=2, fuel=10):
        """Use the Backpack constructor to initialize the name, color,
        and max_size attributes. Fuel is stored separately. A jetpack
        only only 2 item by default.

        Parameters:
            name (str): the name of the jetpack's owner.
            color (str): the color of the jetpack.
            max_size(int): the maximum number of items that can fit inside.
            fuel (int): the amount of fuel left in the jetpack.
        """
        Backpack.__init__(self, name, color, max_size)
        self.fuel = fuel

    def fly(self, amount):
        """Spends a certain amount of fuel to fly. Prints Not enough Fuel!
        when the amount exceeds the fuel."""
        if amount > self.fuel:
            print("Not enough fuel!")
        else:
            self.fuel -= amount

    def dump(self):
        """Resets the contents of the jetpack and empies all fuel."""
        self.contents = []
        self.fuel = 0
````

````python
def test_prob2():
    testpack = Jetpack("Mann", "blue", max_size=4, fuel=15)
    testpack.put("grenade")
    print("Name:", testpack.name)
    print("Color:", testpack.color)
    print("Max size:", str(testpack.max_size))
    print("Fuel:", str(testpack.fuel))
    print("Contents:", str(testpack.contents))
    testpack.fly(10)
    print("Fuel after flying:", str(testpack.fuel))
    testpack.fly(10)
    testpack.dump()
    print("Fuel:", str(testpack.fuel))
    print("Contents:", str(testpack.contents))

test_prob2()
````

````python
jet = Jetpack("Mann", "blue")
assert (jet.max_size, jet.fuel) == (2, 10) and isinstance(jet, Backpack)
jet.fly(4);  assert jet.fuel == 6
jet.fly(7);  assert jet.fuel == 6       # Too much: prints a warning, keeps the fuel.
jet.put("map"); jet.dump()
assert jet.contents == [] and jet.fuel == 0
````

## Magic Methods

A *magic method* is a special method, beginning and ending with two underscores like `__init__()`, that makes an object behave like a built-in data type. Every object has several automatically; in IPython, `b.` + Tab lists the regular methods and attributes, and `b.__` + Tab the magic methods and hidden attributes (`__add__`, `__eq__`, `__lt__`, `__str__`, …).

> **Note.** Many programming languages distinguish between *public* and *private* variables. In Python, all attributes are public, period. However, attributes that start with an underscore are hidden from the user, which is why magic methods do not show up at first in tab completion.

The common magic methods define how an object behaves under binary operations. For backpacks, a simple choice of addition is to add the number of contents, so `A + B` is 8 if `A` has 3 items and `B` has 5. Using `+` calls `__add__()`, with the left object passed as `self` and the right as `other`. Comparisons work the same way: `__lt__()` defines `<`, here "fewer items". Comparison methods should return `True` or `False`, while methods like `__add__()` may return a number or another kind of object.

| Method | Arithmetic Operator | | Method | Comparison Operator |
|---|---|---|---|---|
| `__add__()` | `+` | | `__lt__()` | `<` |
| `__sub__()` | `-` | | `__le__()` | `<=` |
| `__mul__()` | `*` | | `__gt__()` | `>` |
| `__pow__()` | `**` | | `__ge__()` | `>=` |
| `__truediv__()` | `/` | | `__eq__()` | `==` |
| `__floordiv__()` | `//` | | `__ne__()` | `!=` |

What each of these operations does, or *should* do, is up to the programmer and should be carefully documented. See https://docs.python.org/3/reference/datamodel.html#special-method-names.

````python
pack1, pack2 = Backpack("Rose", "red"), Backpack("Carly", "cyan")
pack1.put("textbook")
pack2.put("water bottle")
pack2.put("snacks")
print(pack1 + pack2)                    # Equivalent to pack1.__add__(pack2).

pack1, pack2 = Backpack("Maggy", "magenta"), Backpack("Yolanda", "yellow")
print(pack1 < pack2)
pack2.put('pencils')
print(pack1 < pack2)                    # Equivalent to pack1.__lt__(pack2).
````

## Problem 4: `ComplexNumber`

> **Problem 4.** Write a `ComplexNumber` class from scratch.
> 1. Complex numbers are denoted $a + bi$ where $a, b \in \mathbb{R}$ and $i = \sqrt{-1}$. Write the constructor so it accepts two numbers. Store the first as `self.real` and the second as `self.imag`.
> 2. The *complex conjugate* of $a + bi$ is defined as $\overline{a + bi} = a - bi$. Write a `conjugate()` method that returns the object's complex conjugate as a new `ComplexNumber` object.
> 3. Add the following magic methods:
>    1. Implement `__str__()` so that $a + bi$ is printed out as `(a+bj)` for $b \geq 0$ and `(a-bj)` for $b < 0$.
>    2. The magnitude of $a + bi$ is $|a + bi| = \sqrt{a^2 + b^2}$. The `__abs__()` magic method determines the output of the built-in `abs()` function (absolute value). Implement `__abs__()` so that it returns the magnitude of the complex number.
>    3. Implement `__eq__()` so that two `ComplexNumber` objects are equal if and only if they have the same real and imaginary parts.
>    4. Implement `__add__()`, `__sub__()`, `__mul__()`, and `__truediv__()` appropriately. Each of these should return a new `ComplexNumber` object.
>
> Write a function to test your class by comparing it to Python's built-in `complex` type.
>
> ```python
> def test_ComplexNumber(a, b):
>     py_cnum, my_cnum = complex(a, b), ComplexNumber(a, b)
>     # Validate the constructor.
>     if my_cnum.real != a or my_cnum.imag != b:
>         print("__init__() set self.real and self.imag incorrectly")
>     # Validate conjugate() by checking the new number's imag attribute.
>     if py_cnum.conjugate().imag != my_cnum.conjugate().imag:
>         print("conjugate() failed for", py_cnum)
>     # Validate __str__().
>     if str(py_cnum) != str(my_cnum):
>         print("__str__() failed for", py_cnum)
>     # ...
> ```

````python
class ComplexNumber:

    def __init__(self, real, imag):

        self.real = real
        self.imag = imag

    def conjugate(self):
        conj = ComplexNumber(self.real, -self.imag)
        return conj

    def __str__(self):
        if self.imag >= 0:
            return "("+str(self.real)+"+"+str(self.imag)+"j)"
        else:
            return "("+str(self.real)+str(self.imag)+"j)"

    def __abs__(self):
        from math import sqrt
        return sqrt(self.real**2 + self.imag**2)

    def __eq__(self, other):
        return self.real == other.real and self.imag == other.imag

    def __add__(self, other):
        return ComplexNumber(self.real + other.real, self.imag + other.imag)

    def __sub__(self, other):
        return ComplexNumber(self.real - other.real, self.imag - other.imag)

    def __mul__(self, other):
        r = self.real * other.real - self.imag * other.imag
        i = self.real * other.imag + self.imag * other.real
        return ComplexNumber(r, i)

    def __truediv__(self, other):
        d = other.real**2 + other.imag**2
        r = (self.real*other.real + self.imag*other.imag)/d
        i = (self.imag*other.real - self.real*other.imag)/d
        return ComplexNumber(r, i)
````

````python
def test_prob4(a, b):
    py_cnum, cnum = complex(a,b), ComplexNumber(a, b)
    py_cnum2, cnum2 = complex(a+1, b+3), ComplexNumber(a+1, b+3)
    conj = cnum.conjugate()
    print(cnum)
    print(conj)
    print(abs(cnum))
    print(cnum + cnum2)
    print(cnum - cnum2)
    print(cnum * cnum2)
    print(cnum / cnum2)

    # Validate the constructor.
    if cnum.real != a or cnum.imag != b:
        print("__init__() set self.real and self.imag incorrectly")
    # Validate conjugate() by checking the new number's imag attribute.
    if py_cnum.conjugate().imag != cnum.conjugate().imag:
        print("conjugate() failed for", py_cnum)
    # Validate __str__().
    if str(py_cnum) != str(cnum):
        print("__str__() failed for", py_cnum)
    # Validate __eq__().
    if py_cnum != cnum:
        print("__eq__() failed for", py_cnum)
    # Validate __add__().
    if py_cnum+py_cnum2 != cnum+cnum2:
        print("__add__() failed for", py_cnum)
    # Validate __sub__().
    if py_cnum-py_cnum2 != cnum-cnum2:
        print("__sub__() failed for", py_cnum)
    # Validate __mul__().
    if py_cnum*py_cnum2 != cnum*cnum2:
        print("__mul__() failed for", py_cnum)
    # Validate __truediv__().
    if py_cnum/py_cnum2 != cnum/cnum2:
        print("__truediv__() failed for", py_cnum)

test_prob4(3, 4)
````

The `__truediv__()` "failure" is floating-point rounding, not a wrong formula. `__truediv__()` uses the textbook formula

$$\frac{a + bi}{c + di} = \frac{(ac + bd) + (bc - ad)\,i}{c^2 + d^2},$$

while CPython's built-in `complex` division uses a scaled algorithm to avoid overflow, so the last digit can round differently and the test's exact `!=` flags it. Comparing with a tolerance instead, over a grid of values (every operation, including division, with `cmath.isclose`):

````python
import cmath

def as_complex(z):
    return complex(z.real, z.imag)

checked = 0
for a in range(-5, 6):
    for b in range(-5, 6):
        z, w = complex(a, b), complex(a + 1, b + 3)
        Z, W = ComplexNumber(a, b), ComplexNumber(a + 1, b + 3)
        assert Z.real == a and Z.imag == b
        assert as_complex(Z.conjugate()) == z.conjugate()
        assert str(Z) == str(z) or a == 0       # complex prints a pure imaginary as "4j".
        assert abs(Z) == abs(z) and Z == z
        assert as_complex(Z + W) == z + w and as_complex(Z - W) == z - w
        assert as_complex(Z * W) == z * w
        if w != 0:
            assert cmath.isclose(as_complex(Z / W), z / w, rel_tol=1e-15)
        checked += 1
print(f"{checked} pairs checked")

q_mine, q_py = ComplexNumber(3, 4) / ComplexNumber(4, 7), complex(3, 4) / complex(4, 7)
print(q_mine, q_py, abs(as_complex(q_mine) - q_py))
````

### Static attributes and methods

Attributes accessed through `self` are *instance attributes*, bound to one instance. A *static attribute* is shared between all instances: declare it inside the `class` block but outside any method, without `self`. It can be accessed or changed via the class name without instantiating the class at all.

A *static method* cannot depend on the attributes of individual instances, so there is no `self` in its body or its parameters; it only has access to static attributes and other static methods. Tag it with `@staticmethod`. The spec adds these to `Backpack`; here they go on a subclass so the solution class above stays as written.

````python
class BrandedBackpack(Backpack):
    brand = "Adidas"                    # A static attribute.

    @staticmethod
    def origin():                       # Do not use 'self' as a parameter.
        print("Manufactured by " + BrandedBackpack.brand + ", inc.")

pack1, pack2 = BrandedBackpack("Bill", "blue"), BrandedBackpack("William", "white")
print(pack1.brand, pack2.brand, BrandedBackpack.brand)
BrandedBackpack.brand = "Nike"          # Changes it for all instances.
print(pack1.brand, pack2.brand, BrandedBackpack.brand)
BrandedBackpack.origin()                # No instance needed...
pack1.origin()                          # ...but instances can call it too.
````

### More magic methods

These are particularly important for custom data structure classes:

| Method | Operation | Trigger Function |
|---|---|---|
| `__bool__()` | Truth value | `bool()` |
| `__len__()` | Object length or size | `len()` |
| `__repr__()` | Object representation | `repr()` |
| `__getitem__()` | Indexing and slicing | `self[index]` |
| `__setitem__()` | Assignment via indexing | `self[index] = x` |
| `__iter__()` | Iteration over the object | `iter()` |
| `__reversed__()` | Reverse iteration over the object | `reversed()` |
| `__contains__()` | Membership testing | `in` |

### Hashing

A *hash value* is an integer that uniquely identifies an object; `hash()` computes it by calling `__hash__()`. Python's `set` and `dict` use hash values to store and retrieve objects quickly, so an *unhashable* object can't be put in a set or used as a dictionary key (see https://docs.python.org/3/glossary.html#term-hashable).

If `__hash__()` is not defined, the default hash value is the object's memory address (from `id()`) divided by 16, rounded down. However, two objects that compare equal via `__eq__()` must have the same hash value. (Defining `__eq__()` without `__hash__()`, as `Backpack` does, makes the class unhashable.) A simple conforming `__hash__()` for `Backpack`, using the bitwise XOR operator `^` (`&` and `|` are also good choices):

````python
try:
    {Backpack("Ken", "red")}
except TypeError as e:
    print("TypeError:", e)

class HashableBackpack(Backpack):
    def __hash__(self):
        return hash(self.name) ^ hash(self.color) ^ hash(len(self.contents))

a, b = HashableBackpack("Ken", "red"), HashableBackpack("Ken", "red")
print(a == b, hash(a) == hash(b), len({a, b}))
````

See https://docs.python.org/3/reference/datamodel.html#object.__hash__ for more on hashing.
